## 👤 Select Persona

Choose the persona YAML file to generate a tailored news brief.


```python
import os

PERSONA_DIR = "/content/Gener8/personas"

persona_files = [
    f for f in os.listdir(PERSONA_DIR)
    if f.endswith(".yaml")
]

persona_names = [
    f.replace(".yaml", "").replace("_", " ").title()
    for f in persona_files
]

persona_map = dict(zip(persona_names, persona_files))

persona_names
```

```python
import ipywidgets as widgets
from IPython.display import display

persona_dropdown = widgets.Dropdown(
    options=persona_names,
    description="Select Persona:",
    style={"description_width": "initial"}
)

display(persona_dropdown)
```

```python
from src.persona_loader import load_persona

def load_selected_persona():
    selected_name = persona_dropdown.value
    persona_file = persona_map[selected_name]
    persona_path = f"{PERSONA_DIR}/{persona_file}"
    return load_persona(persona_path)
```

## 🌐 Fetch Articles from RSS Feeds

Articles are pulled from open news sources for the past 30 days.


```python
RSS_FEEDS = [
    "https://techcrunch.com/feed/",
    "https://www.theverge.com/rss/index.xml"
]

articles = []
for feed in RSS_FEEDS:
    articles.extend(fetch_articles(feed, days_back=30))

print(f"Total articles fetched: {len(articles)}")
```

## 🎯 Filter Articles by Persona Relevance


```python
persona = load_selected_persona()
filtered_articles = filter_articles(articles, persona['filters'])
print(f"Relevant articles for persona: {len(filtered_articles)}")
```

## 🧠 Generate Persona-Based Summary


```python
summary = summarize_articles(filtered_articles)
print(summary)
```

## 📰 Generate Final News Brief


```python
brief = generate_brief(persona, summary, filtered_articles)
print(brief)
```
